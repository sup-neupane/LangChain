```python
from langchain_google_genai import GoogleGenerativeAIEmbeddings
from langchain_chroma import Chroma
```

```python
from google.colab import userdata
api_key = userdata.get('GOOGLE_API_KEY')
```

```python
embeddings = GoogleGenerativeAIEmbeddings(
    model="models/gemini-embedding-001",
    google_api_key=api_key
)
```

```python
from langchain_core.documents import Document
# Create LangChain documents for swimmers

doc1 = Document(
    page_content="Michael Phelps is one of the most successful swimmers in Olympic history. Known for his incredible speed, endurance, and versatility, he has won a record number of Olympic gold medals.",
    metadata={"stroke": "Butterfly and Freestyle"}
)

doc2 = Document(
    page_content="Katie Ledecky is one of the greatest female swimmers in history. She is known for her dominance in long-distance freestyle events and exceptional endurance.",
    metadata={"stroke": "Freestyle"}
)

doc3 = Document(
    page_content="Caeleb Dressel is an American competitive swimmer known for his explosive speed and powerful performances in sprint events. He has won multiple Olympic gold medals.",
    metadata={"stroke": "Freestyle and Butterfly"}
)

doc4 = Document(
    page_content="Adam Peaty is one of the fastest breaststroke swimmers in the world. He is known for his powerful technique and multiple world records in breaststroke events.",
    metadata={"stroke": "Breaststroke"}
)

doc5 = Document(
    page_content="Summer McIntosh is a Canadian swimming sensation known for her versatility across multiple swimming events. She has achieved remarkable success in butterfly, freestyle, and individual medley competitions.",
    metadata={"stroke": "Butterfly, Freestyle and Individual Medley"}
)
```

```python
docs = [doc1, doc2, doc3, doc4, doc5]
```

```python
vector_store = Chroma(
    embedding_function=embeddings,
    persist_directory='my_chroma_db',
    collection_name='sample'
)
```

```python
vector_store.add_documents(docs)
```

```python
vector_store.get(include=['embeddings','documents', 'metadatas'])
```

```python
vector_store.similarity_search(
    query="Who among these is female",
    k=2
)
```

```python
vector_store.similarity_search_with_score(
    query='Who among these is female',
    k=2
)
```
